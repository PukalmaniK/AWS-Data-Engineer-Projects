### Pillar 3: Error Handling and Logging
In production, `print()` is not enough. When a pipeline runs at 3 AM, you need robust error handling (`try/except`) and formatted `logging` to trace what went wrong.

💡 **What this code does:**
- Configures the built-in `logging` module to format messages with timestamps and severity levels (INFO, ERROR, CRITICAL).
- Uses a `try/except` block to safely attempt an operation (division).
- Catches specific errors (like `ZeroDivisionError`) to prevent crashes and logs the exact failure details.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.

**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
import logging

# Configure standard Python logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

def safe_divide(a, b):
    try:
        result = a / b
        logger.info(f"Division successful: {result}")
        return result
    except ZeroDivisionError as e:
        logger.error(f"Data Error: Attempted to divide by zero. Details: {e}")
        return None
    except Exception as e:
        logger.critical(f"Unexpected system failure: {e}")
        raise

# Test our robust function
safe_divide(10, 2)
safe_divide(10, 0) # This will gracefully log an error instead of crashing the script
````

### Pillar 5: Building Reusable Modules
Instead of writing messy, procedural scripts, Data Engineers wrap logic into reusable functions or Classes so they can be imported and executed cleanly in pipelines.

💡 **What this code does:**
- Defines a Python `Class` (`S3Utility`) to bundle logic and state (like the bucket name) together.
- Uses an `__init__` method to set up the object and its own logger.
- Provides a reusable method `format_s3_path` to standardize how S3 URIs are generated safely.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.

**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
class S3Utility:
    def __init__(self, bucket_name):
        self.bucket = bucket_name
        self.logger = logging.getLogger(__name__)
        
    def format_s3_path(self, file_name):
        path = f"s3://{self.bucket}/{file_name}"
        self.logger.info(f"Formatted path: {path}")
        return path

# Using our module
utility = S3Utility("my-production-bucket")
utility.format_s3_path("daily_report.csv")
````

# Activity 3: Intro to Boto3 & Programmatic IAM Policies

Now we formally introduce **Boto3**. Boto3 is the official AWS SDK for Python. It allows Python to act like a remote control for your AWS account.

In this step, we will use your unique identifier to generate a specific IAM security policy for your script.

💡 **What this code does:**
- Uses the `my_unique_id` variable you defined earlier.
- Initializes the Boto3 `iam` and `sts` clients to talk to AWS security services.
- Defines a JSON document giving our script permissions to use S3 and CloudWatch Logs.
- Programmatically creates this policy in your AWS account.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.

**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
import boto3, json
from botocore.exceptions import ClientError

# Initialize our Boto3 Clients
iam = boto3.client("iam")
sts = boto3.client("sts")

policy_name = f"boto3-S3logs-{my_unique_id}-Policy"
account_id = sts.get_caller_identity()["Account"]

# Define the IAM Policy Document (JSON rules for what this script is allowed to do)
policy_doc = {
    "Version": "2012-10-17",
    "Statement": [{
        "Effect": "Allow",
        "Action": ["s3:*", "logs:*"],
        "Resource": "*"
    }]
}

try:
    # Programmatically create the policy
    res = iam.create_policy(
        PolicyName=policy_name,
        PolicyDocument=json.dumps(policy_doc)
    )
    print("INFO: IAM Policy Created Successfully.")
    print("ARN:", res["Policy"]["Arn"])

except ClientError as e:
    if e.response["Error"]["Code"] == "EntityAlreadyExists":
        print("INFO: IAM Policy already exists:")
        print(f"ARN: arn:aws:iam::{account_id}:policy/{policy_name}")
    else:
        print("ERROR:", e)
````

````python
import boto3
import json
from botocore.exceptions import ClientError

# We dynamically construct your bucket name based on the ID you provided earlier
TARGET_BUCKET = f"s3-pyboto3-bucket-{my_unique_id}"

s3_client = boto3.client('s3', region_name='us-east-1')

try:
    # Step 1: Disable Block Public Access using Boto3
    s3_client.put_public_access_block(
        Bucket=TARGET_BUCKET,
        PublicAccessBlockConfiguration={
            'BlockPublicAcls': False,        # Allow public ACLs
            'IgnorePublicAcls': False,       # Don't ignore public ACLs
            'BlockPublicPolicy': False,      # IMPORTANT - Allow policies
            'RestrictPublicBuckets': False   # Allow public bucket
        }
    )
    print("INFO: ✓ Disabled Block Public Access via Boto3")

    # Step 2: Create and attach public read policy
    bucket_policy = {
        "Version": "2012-10-17",
        "Statement": [
            {
                "Effect": "Allow",
                "Principal": "*",
                "Action": "s3:GetObject",
                "Resource": f"arn:aws:s3:::{TARGET_BUCKET}/*"
            }
        ]
    }

    s3_client.put_bucket_policy(
        Bucket=TARGET_BUCKET,
        Policy=json.dumps(bucket_policy)
    )
    print("INFO: ✓ Policy attached successfully to", TARGET_BUCKET)

except ClientError as e:
    print("ERROR:", e)
````

### Step 5.1: Set up CloudWatch Logging Functions
A professional script shouldn't just `print()` results; it should log them to the cloud. We will define functions to push logs to Amazon CloudWatch.

💡 **What this code does:**
- Connects to the AWS `logs` service.
- Checks if a specific log group (`/aws/custom/s3-utility-logs`) exists, and creates it if not.
- Defines a reusable function `send_log_to_cloudwatch` that timestamps our messages and pushes them securely to AWS.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.
>
**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
import time
logs_client = boto3.client('logs', region_name='us-east-1')

LOG_GROUP = "/aws/custom/s3-utility-logs"
LOG_STREAM = "ExecutionStream"

def setup_cloudwatch_logs():
    """Create Log Group and Stream in CloudWatch if they don't exist."""
    try:
        logs_client.create_log_group(logGroupName=LOG_GROUP)
    except ClientError: pass # Ignore if it already exists

    try:
        logs_client.create_log_stream(logGroupName=LOG_GROUP, logStreamName=LOG_STREAM)
    except ClientError: pass

def send_log_to_cloudwatch(message):
    """Print locally and send log message to AWS CloudWatch Logs."""
    print(f"INFO: {message}")
    timestamp = int(round(time.time() * 1000))
    try:
        logs_client.put_log_events(
            logGroupName=LOG_GROUP,
            logStreamName=LOG_STREAM,
            logEvents=[{'timestamp': timestamp, 'message': message}]
        )
    except Exception as e:
        print(f"Failed to send log to CloudWatch: {e}")

setup_cloudwatch_logs()
send_log_to_cloudwatch("CloudWatch setup complete. Ready for pipeline execution.")
````

### Step 5.2: Read Data for Upload
We will write a function specifically for reading and validating our CSV file before we let Boto3 touch it.

💡 **What this code does:**
- Opens the local CSV file and extracts all data into a list.
- Uses our new `send_log_to_cloudwatch` function to log the exact row count directly to AWS.
- Gracefully catches a `FileNotFoundError` if the path is wrong.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.
>
**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
import os
import csv

def read_local_csv(file_path):
    """Reads a local CSV file and logs the action."""
    try:
        with open(file_path, mode='r', encoding='utf-8') as file:
            reader = csv.reader(file)
            data = list(reader)
            send_log_to_cloudwatch(f"Successfully read {len(data)} rows from {file_path}")
            return data
    except FileNotFoundError:
        send_log_to_cloudwatch(f"Error: The file {file_path} was not found.")
        return None

local_csv_path = os.path.expanduser("~/Desktop/Project/data.csv")
dataset = read_local_csv(local_csv_path)
````

### Step 5.3: Upload to S3
Now we push the local file to the cloud bucket. 

💡 **What this code does:**
- Uses `s3_client.upload_file()` to transfer the physical file from your local workspace to AWS S3.
- Dynamically names the file `uploads/data.csv` so it organizes neatly into an "uploads" folder in the bucket.
- Logs the success or failure to CloudWatch.
> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.
>
**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
def upload_file_to_s3(file_path, target_bucket, object_name):
    """Uploads the dataset to your S3 bucket."""
    try:
        s3_client.upload_file(file_path, target_bucket, object_name)
        send_log_to_cloudwatch(f"File successfully uploaded to {target_bucket}/{object_name}")
    except ClientError as e:
        send_log_to_cloudwatch(f"Upload failed: {e}")

s3_destination_key = "uploads/data.csv"

# Execute the upload!
upload_file_to_s3(local_csv_path, TARGET_BUCKET, s3_destination_key)
````

### Step 5.4: Verify the Upload via Boto3
Finally, we query S3 to list its contents and prove our file arrived safely.

💡 **What this code does:**
- Uses `s3_client.list_objects_v2` to ask AWS what is currently inside our bucket.
- Iterates through the response and logs the name and size of each file it finds.

> **NOTE:** Please ensure you complete and execute the code cell below without changing the function name, as your results will be validated and automatically reflected in your Practice Progress Summary to track your learning milestones.
>
**⚡ How to run the code:**
Select the below code cell and press **Shift + Enter** or click the **Play (▶)** button on the top left side of the code block to run the code.

````python
def list_s3_objects(target_bucket, prefix=""):
    """Lists objects in the S3 bucket to verify the upload."""
    try:
        response = s3_client.list_objects_v2(Bucket=target_bucket, Prefix=prefix)
        if 'Contents' in response:
            for obj in response['Contents']:
                send_log_to_cloudwatch(f"Found object in S3: {obj['Key']} (Size: {obj['Size']} bytes)")
        else:
            send_log_to_cloudwatch("No objects found with the specified prefix.")
    except ClientError as e:
        send_log_to_cloudwatch(f"Listing objects failed: {e}")

# Verify listing in the 'uploads/' folder
list_s3_objects(TARGET_BUCKET, prefix="uploads/")

send_log_to_cloudwatch("S3 Utility script execution completed successfully.")
````
